```python
def sentimentalAnalysis(data,columnname):
    import matplotlib.pyplot as plt
    import seaborn as sns
    from nltk.sentiment.vader import SentimentIntensityAnalyzer
    sid=SentimentIntensityAnalyzer()
    dataset.dropna(inplace=True)

    data['scores']=data[columnname].apply(lambda commonText:sid.polarity_scores(commonText))
    data['compound']=data['scores'].apply(lambda score_dict:score_dict['compound'])
    data['Negative']=data['scores'].apply(lambda score_dict:score_dict['neg'])
    data['Positive']=data['scores'].apply(lambda score_dict:score_dict['pos'])
    data['Neutral']=data['scores'].apply(lambda score_dict:score_dict['neu'])

    data['comp_score']=data['compound'].apply(lambda c: 'pos' if c>=0 else "neg")
    plt.clf()

    posneg=pd.DataFrame(data['comp_score'].value_counts())
    return posneg,data
```

```python
pos,data_Senti=sentimentalAnalysis(dataset,columnname='Chat')
```

Topic Modelling
-->This is unsupervised learning.
-->group the max topic from the input (pdf file, chat) and give the result


```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
```

```python
tfidf=TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm=tfidf.fit_transform(df["Chat"])
```

```python
from sklearn.decomposition import NMF  #Non matrix Factorization from decomposition method.
nmf_model=NMF(n_components=5,random_state=42)  #n_components to select the topic
nmf_model.fit(dtm)
```

```python
tfidf.get_feature_names_out()
```

```python
for index,topic in enumerate(nmf_model.components_):  #this line creates the nmf_model components in the variable topic with index
    results=([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-10:]])  
    print(results)
```

```python
topic_result=nmf_model.transform(dtm)
dataset["Topic"]=topic_result.argmax(axis=1)
```

```python
dataset
```

```python
dataset["Topic"].value_counts()
```

```python
#LDA (LatentDirichletAllocation) using TF_IDF
#used to find the hidden topics in a collection of txt documents.

from sklearn.decomposition import LatentDirichletAllocation  
lda_model=LatentDirichletAllocation(n_components=5,random_state=42)
lda_model.fit(dtm)
for index,topic in enumerate(lda_model.components_):  
    results=([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-10:]])  
    print(results)
```

```python
topic_result=lda_model.transform(dtm)
dataset["Topic"]=topic_result.argmax(axis=1)
```

```python
def count_vector(data,columnname,vectorizer,topicmodel):
    from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
    from sklearn.decomposition import NMF
    from sklearn.decomposition import LatentDirichletAllocation  
    vector=vectorizer(max_df=0.95, min_df=2,stop_words='english')
    vec=vector.fit_transform(data[columnname])
      
    model=topicmodel(n_components=5,random_state=42)
    model.fit(vec)
    for index,topic in enumerate(model.components_):  
        results=([vector.get_feature_names_out()[i] for i in topic.argsort()[-10:]])  
        print(f"Topic{index}:{results}")
    topic_result=model.transform(vec)
    data["Topic"]=topic_result.argmax(axis=1)
    counts=data["Topic"].value_counts()
    
    return data,counts
```

```python
dataset_topic,count=count_vector(dataset,columnname="Chat",vectorizer=CountVectorizer,topicmodel=NMF)
print(count)
```

```python
dataset_topic,count=count_vector(dataset,columnname="Chat",vectorizer=CountVectorizer,topicmodel=LatentDirichletAllocation)
print(count)
```

```python
#hashing vectorizer  uses the hashing trick to find the token string name to feature integer index mapping.

def hash_vector(data,columnname,vectorizer,topicmodel):
    from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
    from sklearn.decomposition import NMF
    from sklearn.decomposition import LatentDirichletAllocation  
    vector=vectorizer(n_features=1000,ngram_range=(1,2),stop_words='english', alternate_sign=False)
    vec=vector.fit_transform(data[columnname])
      
    model=topicmodel(n_components=5,random_state=42)
    model.fit(vec)
    
    topic_result=model.transform(vec)
    data["Topic"]=topic_result.argmax(axis=1)
    counts=data["Topic"].value_counts()
    
    return data,counts
```

```python
dataset_topic,count=hash_vector(dataset,columnname="Chat",vectorizer=HashingVectorizer,topicmodel=NMF)
print(count)
```

```python
dataset_topic  #vectorizer=HashingVectorizer, topic_model=NMF
```

```python
dataset_topic,count=hash_vector(dataset,columnname="Chat",vectorizer=HashingVectorizer,topicmodel=LatentDirichletAllocation)
print(count)
```
