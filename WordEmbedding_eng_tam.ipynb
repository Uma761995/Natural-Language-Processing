```python
from tensorflow import keras
from tensorflow.keras.layers import Embedding,LSTM,Dense
from tensorflow.keras.models import Sequential
```

```python
train=pd.read_csv("engtamilTrain.csv")
train=train.drop(["Unnamed: 0"],axis=1)
english_sentences=train["en"]
tamil_sentences=train['ta']
english_sentences=english_sentences.head(1000)
tamil_sentences=tamil_sentences.head(1000)
```

```python
from gensim.models import Word2Vec
from sklearn.decomposition import PCA
from matplotlib import pyplot
```

```python
def sentToken(sentence):
    dataset=sentence
    sentences=[sentence.split() for sentence in dataset]  #splitting sentences as a single word

    print(sentences)

    return sentences
```

```python
engSentence=sentToken(english_sentences)
```

```python
tamSentence=sentToken(tamil_sentences)
```

```python
def ownWordmodel(langsentence,modelname):
    from gensim.models import Word2Vec
    from sklearn.decomposition import PCA  #number of features (dimensions) in a dataset while retaining as much important information as possible.
    import matplotlib.pyplot as plt

    model=Word2Vec(langsentence,min_count=1)
    print(model)

    #save model
    model.save(modelname)

    # fit a 2d PCA model to the vectors
    x=model.wv.vectors  #corrected line ,returns all word embeddings as a matrix.
    pca=PCA(n_components=2)
    result=pca.fit_transform(x)

    plt.scatter(result[:,0],result[:,1])
    words=list(model.wv.index_to_key)  #retrieves all words in the Word2Vec vocabulary.
    for i,word in enumerate (words):
        plt.annotate(word,xy=(result[i,0],result[i,1]))
    plt.show()
```

```python
ownWordmodel(engSentence,"engmodel.bin")
```

```python
ownWordmodel(tamSentence,"tammodel.bin")
```
